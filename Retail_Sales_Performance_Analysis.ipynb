```python
# total sales
total_sales = df['Sales'].sum()   
# total profit
total_profit = df['Profit'].sum() 

print("Total Sales:", round(total_sales,2))
print("Total Profit:", round(total_profit,2))
```

Conclusion: 

The analysis showed that the Superstore generated strong overall sales while maintaining a positive profit. This indicates that the business was performing well during the analyzed period. Monitoring both sales and profit together helps in understanding whether increased sales are actually contributing to business growth and profitability.

```python
#Region wise Sales
region_sales = df.groupby('Region')['Sales'].sum()  
print(region_sales)
```

```python
#Region-wise sales Graph
region_sales.plot(kind='bar', figsize=(8,5))

plt.title("Region-wise Sales")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

Conclusion:

The West Region recorded the highest sales contribution among all regions, whereas the South Region generated the lowest sales. This suggests that the West Region has a stronger market presence, while the South Region may require improved marketing strategies and customer engagement to enhance its performance.

```python
 #Top 5 Customers
top_customers = (
    df.groupby('Customer Name')['Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print(top_customers) 
```

```python
 #Top 5 Customers Graph
top_customers.plot(kind='barh', figsize=(8,5))

plt.title("Top 5 Customers")
plt.xlabel("Sales")

plt.show()   
```

Conclusion:

The top five customers contributed a significant portion of the total revenue. Maintaining strong relationships with these valuable customers through loyalty programs and personalized services can help increase customer retention and support future business growth.

```python
 #Botton 5 Products causing loss
bottom_products = (
    df.groupby('Product Name')['Profit']
    .sum()
    .sort_values()
    .head(5)
)

print(bottom_products)
```

```python
 #Bottom 5 products causing loss (Graph)
bottom_products.plot(kind='barh', figsize=(10,5))

plt.title("Bottom 5 Products (Loss)")
plt.xlabel("Profit")

plt.show() 
```

Conclusion:

The identified products generated negative profits despite contributing to sales. This indicates the need to review their pricing, discount policies, operational costs, and demand patterns. Addressing these issues can help reduce losses and improve overall profitability.

```python
#Proper Order Date 
df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    format='%d-%m-%Y'
)

df.head() 
```

Conclusion:

The Order Date column was converted into the appropriate datetime format to ensure accurate date-based analysis. This transformation enabled the extraction of time-related insights such as monthly trends and year-wise sales growth. Proper date formatting is essential for performing reliable time series analysis and generating meaningful business insights.

```python
 # Monthly Sales 
monthly_sales = df.groupby(
    df['Order Date'].dt.to_period('M')
)['Sales'].sum()

print(monthly_sales) 
```

```python
 #Monthly Sales Trend Graph
monthly_sales.plot(figsize=(12,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show() 
```

Conclusion:

The monthly sales trend showed fluctuations throughout the year, with higher sales observed during certain months. This indicates seasonal demand patterns. Understanding these trends can help businesses plan inventory, promotions, and resource allocation more effectively.

```python
 # Category-wise Sales
category = df.groupby('Category')[['Sales','Profit']].sum()

print(category) 
```

```python
#Category-wise Sales Graph
category['Sales'].plot(kind='bar', figsize=(8,5))

plt.title("Category-wise Sales")
plt.ylabel("Sales")

plt.show()   
```

Conclusion:

Among all product categories, Technology performed the best in terms of sales, while the other categories showed comparatively lower contributions. This suggests that investing more in high-performing categories while improving weaker ones can enhance overall business performance.

```python
 #Segment wise Profit
segment = df.groupby('Segment')[['Sales','Profit']].sum()

segment['Profit Ratio'] = (
    segment['Profit'] /
    segment['Sales']
) * 100

print(segment) 
```

```python
#Segment wise Profit Ratio graph
segment['Profit Ratio'].plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title("Profit Ratio by Segment")
plt.ylabel("")

plt.show() 
```

Conclusion:

The profitability of customer segments varied significantly. Some segments contributed more effectively to profit generation than others. Identifying highly profitable customer segments enables businesses to focus their efforts on customers who create greater value.Here Home Office Category generates the highest Profit Ratio.

```python
 #Average Order Value 
average_order_value = (
    df['Sales'].sum() /
    df['Order ID'].nunique()
)

print(
    "Average Order Value:",
    round(average_order_value,2)
)                                    
```

Conclusion:

The average order value provided insights into customer purchasing behavior and spending capacity. Monitoring this metric helps businesses evaluate customer buying patterns and develop strategies to encourage higher-value purchases.

```python
#Discount vs Profit Impact
high_discount = df[df['Discount'] > 0.30]

impact = high_discount.groupby(
    'Discount'
)['Profit'].mean()

print(impact)   
```

```python
 #correlation of Discount between Profit
plt.show    
impact.plot(kind='bar', figsize=(8,5))

plt.title("High Discount Impact")
plt.xlabel("Discount")
plt.ylabel("Average Profit")

plt.show                      
```

Conclusion:

The analysis revealed that orders with higher discounts often resulted in lower profits. While discounts can increase sales volume, excessive discounting can negatively affect earnings. Therefore, businesses should implement discount strategies carefully to maintain profitability.

```python
#correlation of Discount between Profit
correlation = df[
    ['Discount','Profit']
].corr()

print(correlation) 
```

```python
#correlation of Discount between Profit Graph
plt.figure(figsize=(8,5))

plt.scatter(
    df['Discount'],
    df['Profit']
)

plt.title(
    "Correlation Between Discount and Profit"
)

plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()         
```

Conclusion:

The correlation analysis indicated a negative relationship between discounts and profit. As discount levels increased, profitability tended to decrease. This finding highlights the importance of balancing promotional activities with profit objectives to achieve sustainable business g

```python
#Sales by Ship Mode Graph
ship_mode = df.groupby('Ship Mode')['Sales'].sum()

print(ship_mode)

ship_mode.plot(kind='bar')

plt.title("Sales by Ship Mode")
plt.ylabel("Sales")

plt.show()       
```

Conclusion:

Standard Class recorded the highest sales, making it the most preferred shipping mode among customers. Same Day shipping generated the lowest sales, indicating lower customer demand for faster delivery options. The company should continue focusing on improving its standard shipping services to maintain customer satisfaction and sales growth.

```python
#Profit by Sub-Category graph
subcategory = (
    df.groupby('Sub-Category')['Profit']
    .sum()
    .sort_values()
)

subcategory.plot(kind='barh', figsize=(10,7))

plt.title("Profit by Sub-Category")
plt.xlabel("Profit")

plt.show()                                    
```

Conclusion:

The analysis reveals that Tables and Machines are the biggest loss-making sub-categories, followed by Phones, Storage, and Supplies. These products require immediate attention to identify the reasons behind poor profitability. Revising pricing strategies, controlling costs, and optimizing discounts can help reduce losses and improve overall business performance.

```python
# Year wise Sales Growth Graph
yearly_sales = df.groupby(
    df['Order Date'].dt.year
)['Sales'].sum()

print(yearly_sales)

yearly_sales.plot(marker='o')

plt.title("Year-wise Sales Growth")
plt.xlabel("Year")
plt.ylabel("Sales")

plt.show()
```

Conclusion:
Sales grew significantly from 2014 to 2017, with 2017 recording the highest sales among all years. This upward trend reflects strong business growth and increasing customer engagement. The company should continue the strategies that contributed to this success while identifying opportunities for further expansion.
